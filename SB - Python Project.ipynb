You are an analyst working for a university's career services office. The office wants to understand what predicts whether a student gets placed in a job after their MBA, so they can better advise future students. Answer all of the questions below using Python and pandas.

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\ADMIN\Desktop\Business Analytics Class\Python\Assignment\Campus Placement.csv")
## shape returns number of rows and columns of a data frame. The data set has 215 rows and 15 columns.
df.shape
## info() returns metadata about the data frame; column names, their data types, missing values, 
df.info()
```

```python
df.describe()
## Missing values in the ssalary column are for those students who were not placed.
```

```python
## What percentage of students in this dataset were placed in a job? (hint: value_counts on the status column).
## Formula :- count "Placed" value in status column/ count number of rows * 100
df.head(10)
placed_count = (df["status"] == "Placed").sum()
tot_students = len(df)
percentage_placed = (placed_count/tot_students)*100

print(f"{percentage_placed:.2f}% of students were placed.")
```

```python
##Does having prior work experience (workex) affect placement? Calculate the placement rate for students with work experience vs without.
## What's the difference?
##Solution : Group by those with work experience and those without.
rates = df.groupby('workex')['status'].value_counts(normalize=True) * 100
placed_rates = rates.xs('Placed', level='status')

print(placed_rates.round(2))
```

```python
## Does MBA specialisation affect placement rate? Compare Mkt&HR vs Mkt&Fin.
df.head(15)
##Group data by specialization
placed = (
    df[df['status'] == 'Placed']
    .groupby('specialisation')['status']
    .count()
)
percentage1 = ((placed/tot_students)*100).round(2)
not_placed = (
    df[df['status'] == 'Not Placed']
    .groupby('specialisation')['status']
    .count()
)
print(placed)
print(percentage1)
print(not_placed)
```

```python
##Is there a gender difference in placement rate? Calculate it. Report the number plainly and professionally -- 
## don't speculate about causes you can't see in the data, just state what the numbers show.
df.head(5)
## Group by gender then by placement and calculate the percentage.
gender_placement_rate = df.groupby('gender')['status'].value_counts(normalize=True)*100
print(gender_placement_rate)
```
